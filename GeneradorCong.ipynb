```python
def gnerador_congruencial(a,b,m,X0):
  Xi = (a*X0 + b) %  m
  return Xi
```

```python
n = 10
a = 7
b = 5
m = 32
X0 = 0
muestra = []
for i in range(n):
  nuevo = gnerador_congruencial( a,b,m, X0 = nuevo)
  muestra.append(nuevo)

```
